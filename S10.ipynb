**BackPropagation:**

زمانی مرحله Forward Path انجام می‌شود و داده وارد شیکه شده، در خورجی یک میزان خطا ایجاد می شود که این خطای ایجاد شده برخواسته از مقادیر موجود در وزن ها هست. در این مرحله لازم است که ما این وزن ها را بدانیم به چه میزان در رخداد این خطا سهیم هستند که لازم این میزان این سهم حطا را با استفاده از Back Propagation محاسبه کنیم. در مرحله بعد از اینکه این مزان سهم هر وزن در ایجاد خطا مشخص شد می آییم بر اساس این سهم مقدار های وزن هارا به روزرسانی می کنیم که در این صورت باعث شود میزان خطا کاهش یابد. برای قهم این موضوع چن مثال در 4 سناریو مختلف حل می کنیم:


***مثال 1: Fully Connected ساده***

معرفی سناریو:

2 ورودی

1 لایه پنهان (2 نورون)

1 خروجی

Activation: Sigmoid

Loss: MSE


حل:

Forward Pass (پیش‌رو):


```
y = x × w = 2 × 0.5 = 1
خطا = target - y = 4 - 1 = 3
```

می‌خواهیم بفهمیم: w را چقدر تغییر دهیم؟

ایده: ببینیم اگر w را کمی تغییر دهیم، خطا چقدر تغییر می‌کند.

```

# اگر w = 0.5 → خطا = 3
# اگر w = 0.6 → y = 2×0.6 = 1.2, خطا = 4 - 1.2 = 2.8
# اگر w = 0.7 → y = 1.4, خطا = 2.6

# می‌بینیم: هر بار w را 0.1 زیاد می‌کنیم، خطا 0.2 کم می‌شود
# یعنی: مشتق (نرخ تغییر) = 0.2 / 0.1 = 2
```

به‌روزرسانی:

```
w_new = w + (نرخ یادگیری) × (تأثیر)
w_new = 0.5 + 0.1 × 2 = 0.7
```

خالا در ادامه این مثال یک نمونه با Activation تابع Sigmoid بررسی می شود:

سناریو:

```
x = 1
w = 0.5
b = 0.1
target = 0.8
```

Forward Pass:

```
# مرحله 1: جمع وزین
z = (x × w) + b = (1 × 0.5) + 0.1 = 0.6

# مرحله 2: sigmoid
y = sigmoid(z) = 1 / (1 + e^(-0.6)) = 0.646

# مرحله 3: خطا (MSE)
error = (target - y)² = (0.8 - 0.646)² = (0.154)² = 0.0237
```
حالا می‌خواهیم بدانیم: w را چقدر تغییر دهیم؟

مشکل: خطا مستقیماً به w وابسته نیست. این مسیر را طی کرده:

```
w → z → y → error
```
راه حل: قاعده زنجیره‌ای - از انتها به اول برویم.


مرحله 1: تأثیر خطا روی y

```
derror/dy = 2 × (y - target) = 2 × (0.646 - 0.8) = 2 × (-0.154) = -0.308
```

یعنی: اگر y را 1 واحد زیاد کنم، خطا 0.308 واحد کم می‌شود.


مرحله 2: تأثیر y روی z (مشتق sigmoid)

```
dy/dz = y × (1 - y) = 0.646 × (1 - 0.646) = 0.646 × 0.354 = 0.228
```

یعنی: اگر z را 1 واحد زیاد کنم، y به اندازه 0.228 زیاد می‌شود


مرحله 3: تأثیر z روی w

```
dz/dw = x = 1
```

یعنی: اگر w را 1 واحد زیاد کنم، z به اندازه 1 زیاد می‌شود.


مرحله 4: قاعده زنجیره‌ای - ترکیب همه

```
derror/dw = derror/dy × dy/dz × dz/dw
          = (-0.308) × 0.228 × 1
          = -0.070
```

یعنی: اگر w را 1 واحد زیاد کنم، خطا 0.070 واحد کم می‌شود.

به‌روزرسانی وزن:

```
learning_rate = 0.5
w_new = w - learning_rate × derror/dw
      = 0.5 - 0.5 × (-0.070)
      = 0.5 + 0.035
      = 0.535
```
سناریو جدید:

```
x = 1
w1 = 0.3, b1 = 0.1
w2 = 0.4, b2 = 0.2
target = 0.9
```

Forward Pass:

```
# لایه 1
z1 = x×w1 + b1 = 1×0.3 + 0.1 = 0.4
h = sigmoid(z1) = sigmoid(0.4) = 0.598

# لایه 2
z2 = h×w2 + b2 = 0.598×0.4 + 0.2 = 0.239 + 0.2 = 0.439
y = sigmoid(z2) = sigmoid(0.439) = 0.608

# خطا
error = (target - y)² = (0.9 - 0.608)² = 0.085
```

Backward Pass (از عقب به جلو):

مرحله 1: تأثیر روی w2


```
derror/dy = 2×(y - target) = 2×(0.608 - 0.9) = -0.584
dy/dz2 = y×(1-y) = 0.608×0.392 = 0.238
dz2/dw2 = h = 0.598

derror/dw2 = (-0.584) × 0.238 × 0.598 = -0.083
```

مرحله 2: انتشار خطا به لایه قبل (h)

```
derror/dh = derror/dy × dy/dz2 × dz2/dh
          = (-0.584) × 0.238 × w2
          = (-0.584) × 0.238 × 0.4 = -0.055
```

مرحله 3: تأثیر روی w1

```
dh/dz1 = h×(1-h) = 0.598×0.402 = 0.240
dz1/dw1 = x = 1

derror/dw1 = derror/dh × dh/dz1 × dz1/dw1
           = (-0.055) × 0.240 × 1 = -0.013
```

به‌روزرسانی هر دو وزن:

```
learning_rate = 0.5

w2_new = 0.4 - 0.5 × (-0.083) = 0.4 + 0.0415 = 0.4415
w1_new = 0.3 - 0.5 × (-0.013) = 0.3 + 0.0065 = 0.3065
```

Forward Pass:

مرحله 1: اعمال فیلتر روی تصویر (4 موقعیت)


موقعیت 1 (بالا-چپ): پیکسل‌های [1,2,4,5]


```
z1 = (1×0.1) + (2×0.2) + (4×0.3) + (5×0.4) + b
```

موقعیت 2 (بالا-راست): پیکسل‌های [2,3,5,6]

```
z2 = (2×0.1) + (3×0.2) + (5×0.3) + (6×0.4) + 0.1
```

موقعیت 3 (پایین-چپ): پیکسل‌های [4,5,7,8]

```
z3 = (4×0.1) + (5×0.2) + (7×0.3) + (8×0.4) + 0.1

    = 6.8
```

موقعیت 4 (پایین-راست): پیکسل‌های [5,6,8,9]

```
z4 = (5×0.1) + (6×0.2) + (8×0.3) + (9×0.4) + 0.1
    = 0.5 + 1.2 + 2.4 + 3.6 + 0.1

```

```

[6.8, 7.8]
```

```

```

 انتشار خطا به هر پیکسل Feature map

 ```

 ```

## **معماری CNN + FCN با تمام جزئیات**

## **معماری شبکه:**

```
ورودی: تصویر 4×4 (1 کانال)
↓
Conv2D: 1 فیلتر 3×3, padding='same', stride=1, use_bias=True, activation='relu'
↓
Feature map: 4×4×1
↓
MaxPooling2D: pool_size=2×2, stride=2
↓
Feature map: 2×2×1
↓
Flatten: (4)
↓
Dense: 4 → 3 نورون, activation='relu', use_bias=True
↓
Dense: 3 → 2 نورون (خروجی), activation='softmax', use_bias=True
↓
Loss: Sparse Categorical Crossentropy
```

## **مرحله 5: Dense خروجی (3 → 2) با Softmax**

```

       = 8.316 + 39.756 + 94.32 + 0.1 = 142.492

z_out2 = 83.16×0.4 + 198.78×0.5 + 314.4×0.6 + 0.2
```

### Softmax:
```
e^142.492 ≈ 1.2 × 10^62 (عددی بسیار بزرگ)
e^321.494 ≈ 2.1 × 10^139 (عددی بسیار بسیار بزرگ)

خروجی1 = e^142.492 / (e^142.492 + e^321.494) ≈ 0
خروجی2 = e^321.494 / (e^142.492 + e^321.494) ≈ 1
```

```

```

## **مرحله 7: Backward Propagation (فقط ساختار کلی)**

### مشتق Loss نسبت به خروجی Softmax:
```
dLoss/d(out1) = 0 (چون out1≈0)
dLoss/d(out2) = -1/out2 ≈ -1
```

### انتشار به لایه Dense خروجی:
مشتق نسبت به `z_out1` و `z_out2` با مشتق Softmax (که وابسته به همه خروجی‌هاست)

### انتشار به لایه Dense اول:
از طریق `W_dense2` و مشتق ReLU (که برای مقادیر مثبت = 1)

### انتشار به MaxPooling:
در MaxPooling، خطا فقط به **بزرگترین عنصر در هر پنجره** می‌رسد (بقیه 0)

مثلاً در پنجره اول (41.7, 55.1, 89.3, 102.7)، بزرگترین 102.7 است و خطا به آن می‌رسد.

### انتشار به Conv2D:
از طریق Pooling و سپس از طریق مشتق ReLU (1 برای مقادیر مثبت)

مشتق نسبت به هر وزن Conv2D = جمع مشتقات از **همه موقعیت‌هایی** که آن وزن در آنها ظاهر شده (همانطور که قبلاً تمرین کردید)

### انتشار به Padding:
Padding **هیچ مشتقی ندارد** (اعداد صفر هستند و وزن ندارند)

---

## **جدول خلاصه تأثیر لایه‌ها بر Backpropagation:**

| لایه | تأثیر بر Backpropagation |
|------|-------------------------|
| **Conv2D** | وزن‌ها در چند موقعیت ظاهر می‌شوند → جمع مشتقات |
| **Padding** | فقط داده اضافه می‌کند، وزن ندارد → تأثیر غیرمستقیم |
| **ReLU** | اگر ورودی >0: گرادیان = 1، اگر ≤0: گرادیان = 0 |
| **MaxPooling** | فقط به حداکثر مقدار گرادیان می‌دهد (بقیه 0) |
| **Flatten** | فقط شکل داده را تغییر می‌دهد، گرادیان مستقیماً عبور می‌کند |
| **Dense** | مشابه Fully Connected (ضرب در W^T) |
| **Softmax** | مشتق ماتریسی که وابستگی بین خروجی‌ها را شامل می‌شود |

---

## **نکات کلیدی که باید به خاطر بسپارید:**

1. **Padding='same'**: خروجی Conv هم‌ابعاد ورودی است، اما تصویر با صفر Padding می‌شود
2. **ReLU**: نورون‌های مرده (ورودی ≤0) گرادیان 0 می‌گیرند
3. **MaxPooling**: فقط یک عنصر در هر پنجره گرادیان می‌گیرد (بقیه از یادگیری بازمی‌مانند)
4. **Flatten**: فقط مرتب‌سازی مجدد، تأثیری در گرادیان ندارد
5. **Softmax**: خروجی‌ها به هم وابسته هستند → مشتق آن یک ماتریس است

---

اگر **یک نورون در لایه Dense1** مقدار منفی (مثلاً -5) داشت (قبل از ReLU)، آن نورون چه تأثیری در Backpropagation داشت؟

**پاسخ:** خروجی آن نورون = 0 (چون ReLU)، و گرادیان آن نورون = 0 → وزن‌های متصل به آن نورون **هیچ به‌روزرسانی** نمی‌شدند!

## **`.compile()` در طراحی مدل چیست؟**

---

## پاسخ خیلی کوتاه:

**`.compile()` همان جایی است که شما به مدل می‌گویید "چطور می‌خواهی یاد بگیری؟"**

---

## توضیح قدم به قدم:

### **گام اول: تشبیه ساده**

فرض کنید یک دانش‌آموز (مدل) را آماده کرده‌اید. حالا باید به او بگویید:

| سؤال | در مدل |
|------|--------|
| **چطور اشتباهاتت را اندازه بگیرم؟** | Loss Function |
| **چطور وزن‌هایت را اصلاح کنی؟** | Optimizer |
| **می‌خواهم چطور عملکردت را ببینم؟** | Metrics |

**`.compile()` = به دانش‌آموز گفتن "قوانین یادگیری"**

---

### **گام دوم: سه جزء اصلی `.compile()`**

```python
model.compile(
    optimizer='adam',           # 1. موتور یادگیری
    loss='sparse_categorical_crossentropy',  # 2. خطکش خطا
    metrics=['accuracy']        # 3. نمایشگر عملکرد
)
```

---

### **1. Optimizer (بهینه‌ساز) = موتور یادگیری**

**وظیفه:** تصمیم می‌گیرد **چقدر و در چه جهتی** وزن‌ها را تغییر بدهد.

| Optimizer | توضیح | چه موقع استفاده شود؟ |
|-----------|-------|---------------------|
| `'adam'` | محبوب‌ترین، سریع، خودکار | **تقریباً همیشه** (پیش‌فرض) |
| `'sgd'` | ساده‌ترین | شبکه‌های کوچک، یادگیری تدریجی |
| `'rmsprop'` | خوب برای RNN | داده‌های دنباله‌ای (متن، زمان) |
| `'adagrad'` | خوب برای داده‌های پراکنده | بین‌المللی سازی، داده کم‌تکرار |

**تشبیه:** مثل **راننده ماشین** که تصمیم می‌گیرد چقدر گاز بدهد و فرمان را چه مقدار بچرخاند.

---

### **2. Loss Function (تابع ضرر) = خطکش خطا**

**وظیفه:** اندازه می‌گیرد **پیش‌بینی مدل چقدر با واقعیت فاصله دارد**.

| مشکل | Loss | مثال |
|------|------|------|
| **چندکلاسه** (10 کلاس CIFAR) | `'sparse_categorical_crossentropy'` | گربه، سگ، ماشین، ... |
| **دوکلاسه** (گربه یا سگ) | `'binary_crossentropy'` | تشخیص خوب/بد |
| **پیش‌بینی عدد** (قیمت خانه) | `'mse'` (Mean Squared Error) | رگرسیون |

**تشبیه:** مثل **خطکش** که می‌گوید تیرت چقدر به هدف فاصله دارد.

---

### **3. Metrics (معیارها) = نمایشگر عملکرد**

**وظیفه:** به شما نشان می‌دهد **مدل چقدر خوب کار می‌کند** (برای انسان، نه برای یادگیری).

| Metric | معنی |
|--------|------|
| `'accuracy'` | چند درصد درست جواب داده |
| `'mse'` | میانگین مربعات خطا |
| `'mae'` | میانگین قدر مطلق خطا |

**تشبیه:** مثل **نمره‌دهی** به دانش‌آموز (برای اطلاع شما، نه برای یادگیری او).

---

### **گام سوم: فرق Loss و Metrics**

| | Loss | Metrics |
|---|------|---------|
| **هدف** | راهنمایی مدل برای یادگیری | نمایش عملکرد به انسان |
| **تأثیر** | مستقیماً وزن‌ها را تغییر می‌دهد | هیچ تأثیری در یادگیری ندارد |
| **الزام** | حتماً باید یک Loss داشته باشد | اختیاری است |
| **مثال** | crossentropy | accuracy |

---

### **گام چهارم: انواع کامپایل برای پروژه‌های مختلف**

#### **1. CIFAR-10 (شما - طبقه‌بندی چندکلاسه):**

## **یک نکته بسیار مهم:**

`.compile()` فقط **یک بار** و قبل از `.fit()` اجرا می‌شود.

اگر بخواهید Optimizer یا Loss را عوض کنید، باید دوباره کامپایل کنید:

```python
# کامپایل اول
```
